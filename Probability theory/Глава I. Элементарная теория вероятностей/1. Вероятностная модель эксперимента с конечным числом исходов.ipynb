# 1. 1. Вероятностная модель эксперимента с конечным числом исходов

1. Рассмотрим некоторый эксперимент, результаты которого (при данном «комплексе условий») описываются конечным числом различных исходов (явлений) $\omega_1, ..., \omega_N.$ Для нас несущественна реальная природа этих
исходов, важно лишь то, что их число $N$ конечно.

Исходы $\omega_1, ..., \omega_N.$ также называются элементарными событиями, а их совокупность
$$\Omega = \{\omega_1, ..., \omega_N. \}$$

(конечным) пространством элементарных событий или пространством событий. 

Выделение пространства элементарных событий представляет собой первый шаг в формулировании понятий вероятностной модели какого-то эксперимента.

Выбор с возвращением - называют эксперимент, в котором на каждом шаге извлеченный шар возвращаются обратно. В этом случае каждая выборка из $n$ шагов может быть записана в виде $(a_1, ..., a_n)$, где $a_i - $ номер шара, извлченного на i-м шаге. В случае выбора с возвращением каждое $a_i$ может принимать любое из $M$ значений $1, 2, ..., M$.

Рассмотрим упорядоченные выборки $(a_1, ..., a_n)$  с возвращением пространство элементарных событий $\Omega$ 

$$\Omega = \{\omega: \omega = (a_1, ..., a_n), a_i = 1, ..., M \}$$

и число различных исходов (выборок), называемых в комбинаторике размещениями из $M$ по $n$ с повторениями, равно 

$$N(\Omega) = M^n \tag{1}$$

Если же рассматриваются неупорядоченные выборки с возвращением (сочетание из M по n с повторениями), то 

$$\Omega = \{\omega: \omega = [a_1, ..., a_n] a_i = 1,..., M \}$$
