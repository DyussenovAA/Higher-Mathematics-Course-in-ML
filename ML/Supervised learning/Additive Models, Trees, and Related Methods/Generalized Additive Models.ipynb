# Generalized Additive Models 

Регрессионные модели и их ограничения

Регрессионные модели занимают центральное место в анализе данных. Они нужны не только для предсказаний и классификации, но и для того, чтобы понять, какие входные переменные (предикторы) действительно важны. Однако классическая линейная модель, при всей своей простоте и удобстве, на практике часто оказывается бесполезной. Главная причина — в реальном мире связи между переменными почти никогда не бывают строго линейными. Эффекты могут усиливаться, замедляться или менять знак в зависимости от значений предиктора.

От линейных моделей к гибким методам

Один из способов борьбы с нелинейностью - использование заранее заданных базисных функций (например, полиномов или сплайнов с фиксированными узлами). Это помогает, но требует от исследователя априорного выбора формы кривой. В данном разделе предлагается более автоматизированный и гибкий подход. Они называются обобщёнными аддитивными моделями (generalized additive models, GAM). Эти методы сами "обучаются" форме нелинейных зависимостей прямо по данным.

В этой серии описываются более автоматические гибкие статистические методы, которые могут использоваться для выявления и описания нелинейных регрессионных эффектов. Эти методы называются "обобщёнными аддитивными моделями".

В контексте регрессии обобщенная аддитивная модель имеет вид:

$$E(Y|X_1,X_2, ..., X_p) = \alpha + f_1(X_1) + f_2(X_2) + ... + f_p(X_p).$$

- $Y$ - Зависимая переменная (отклик) которую мы предсказываем

- $X_1, X_2, ..., X_p$ - предикаторы (независимые переменные)
