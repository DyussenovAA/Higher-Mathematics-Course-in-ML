# 1. Подпространства

Всякий базис подпространства $U$ векторного пространства $V$ можно дополнить до базиса всего пространства. Таким образом мы получим базис, в котором данное пространство $U$ выглядит весьма просто, а именно, натянуто на несколько первых базисных векторов.

**Определение 1.** Базис пространства $V$ называется согласованным с подпространством $U$, если $U$ является линейной оболочкой какой-то части базисных векторов (т.е. одним из "координатных подпространств" относительно этого базиса).

Так, на рис. 1, а) базис $\{e_1, e_2\}$ согласован с подпростнаством $U$, а на рис. 1, б) - нет.



Согласно сказанному выше для всякого подпространства существует согласованный с ним базис.

Пусть теперь имеются два подпространства $U$, $W \subset V$. Очевидно, что их пересечение $U \cap W$ также является подпространством. Это наибольшее подпространство, содержащееся как в $U$, так и в $W$.

**Определение 2.** Суммой $U + W$ подпространств $U$ и $W$ называется совокупность векторов вида $u + w$, где $u \in U$, $w \in W$. Это наименьшее полдпространство, содержащеее как $U$, так и $W$. Иными словами, это линейная оболочка объединения $U \cup W$.

**Теорема 1.** Для всякой пары подпростанств $U$, $W \subset V$ существует базис пространства  $V$, согласованный с каждым мз подпространств $U, W$.

**Доказательство**. Пусть $\{e_1, ..., e_p\}$ - базис подпространства $U \cap W$. Дополним его какими-то векторами $e_{p+1}, ..., e_k$ до базиса подпространства $U$ и, с другой стороны, векторами $e_{k + 1}, ..., e_{k + l - p}$ - до базиса подпространства $W$. (Здесь $p = dim U \cap W, k = dim U, l = dim W.$) Докажем, что векторы $e_1, ..., e_{k + l - p}$ линейно независимы. Дополнив их затем до базиса подпространства $V$, мы получим базис, согласованный как с $U$, так и с $W$.

предположим, что 

$$\sum^{k + l - p}_{i = 1} \lambda_i e_i = 0$$

рассмотрим вектор 

$$x = \sum^k_{i=1} \lambda_i e_i = - \sum^{k + l - p}_{i = k + 1} \lambda_i e_i$$

из первого представления вектора $x$ следует, что он лежит в $U$, а из второго - что он лежит в $W$. Таким образом, $x \in U \cap W$ и значит, 

$$x = \sum^p_{i=1} \mu_i e_i = - \sum^{k + l - p}_{i = k + 1} \lambda_i e_i$$

так как векторы $e_1, ..., e_p, e_{k + 1}, ..., e_{k + l - p}$ линейно независимы, то отсюда следует, что $x = 0$ и $\lambda_i = 0$ при $i = k + 1, ..., k + l - p$. Далее, т.к. векторы $e_1, ..., e_k$ линейно независимы, то из равенства 

$$\sum^k_{i=1} \lambda_i e_i = 0$$

следует, что $\lambda_i = 0$ при $i = 1, ...,k$.


Рисунок 2 иллюстрирует это доказательство в случае $p = 1, k = l = 2$.

**Определение 5.** Говорят, что векторное пространство $V$ разлагается в прямую сумму подространств $U_1, ..., U_k$, если 

1) подпространства $U_1, ..., U_k$ линейно независимы
2) $U_1 + ... + U_k = V$
В этом случае пишут 

$$V = U_1 \oplus ... \oplus U_k$$

Каждый вектор $v \in V$ однозначно представляется в виде $v = u_1 + ... + u_k$, где $u_i \in U_i$, вектор $u_i$ называется проекцией вектора $v$ на подпространство $U_i$. Подчеркнем, что проекция вектора на подпространство $U_i$ зависит не только от этого подпространства, но и от остальных слагаемых разложения. 

**Пример 1.** Квадратная матрица $A$ называется симметричной, если $A^T = A$, и кососимметричной, Если $A^T = - A$. Симметричные (соответственно кососимметричные) матрицы образуют подпространство $L_n^+ (K)$ (соответственно $L_n^-(K)$) в пространстве $L_n (K)$ всех матриц. При условии, что $char K \neq 2$, всякая матрица $A$ может быть представлена в виде суммы симметричной и коссимметричной матриц:

$$A = \frac{1}{2}(A + A^T) + \frac{1}{2}(A - A^T)$$

С другйо стороны, при том же условии очевидно, что матрица, симметричная и кососимметричная одновременно, равна нулю. Это означает, что 

$$L_n (K) = L^+_n(K) \oplus L_n^-(K).$$

**Пример 2.** Аналогично доказывается, что пространство всех функций не вещественной прямой является прямой суммой подпространств четных и нечетных функций. (В жтом рпимере векторное пространство и оба подпространства бесконечномерны.)
