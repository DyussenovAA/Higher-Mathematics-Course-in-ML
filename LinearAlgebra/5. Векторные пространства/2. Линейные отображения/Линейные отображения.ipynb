# 2. Линейные отображения 

Подобно гомоморфизмам групп естественно рассматривать го-
моморфизмы векторных пространств. Их принято называть линей-
ными отображениями.

**Определение 1.** Пусть $U$ и $V$ - векторные пространства над полем $K$. Отображение 

$$\phi: V \to U$$

называется линейным, если 

1) $\phi (a + b) = \phi(a) + \phi(b)$ для любых $a, b \in V$ 

2) $\phi(\lambda a) = \lambda \phi(a)$ для любых $\lambda \in K, a \in V$
