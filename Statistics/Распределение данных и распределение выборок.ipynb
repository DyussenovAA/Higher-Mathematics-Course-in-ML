## Стандартная ошибка

**Стандартная ошибка** - это одиночная метрика, которая обобщает вариабельность статистической величины в выборочном распределении. Стандартную ошибку можно оценить с использованием стастической величины, опираясь на стандартное отклонение $s$ значений выорки и размерк выборки $n$: 

$$\text{стандартная ошибка} = \frac{s}{\sqrt{n}}$$
