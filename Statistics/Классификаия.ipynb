## Ковариация 

Мера степени, с которой переменная верьируется с другой (т.е. с похожей магнитудой и направлением)

Ковариация измеряет связь между двумя переменными - $x, z$. Ковариция медлу $x,z$ задается как:

$$s_{x,z} = \frac{\sum_{i=1}^{n}(x_i - \overline{x})(z_i - \overline{z})}{n-1}$$

где $n$ - число записей. см. почему n-1?
